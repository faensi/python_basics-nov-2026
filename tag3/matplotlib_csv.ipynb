## Linie zeichnen

`matplotlib.pyplot` stellt die Figur. `plot` zeichnet eine Linie. Mehrere Aufrufe von `plot` legen mehrere Reihen in dieselbe Figur. `xlabel`, `ylabel` und `title` beschriften die Achsen. `legend` zeigt die Namen der Reihen. `show` zeigt die Figur.


```python
import pandas as pd
import matplotlib.pyplot as plt

table = pd.read_csv("data/pump_pressure.csv")
inlet = table[table["port"] == "inlet"]["pressure_kpa"].to_numpy()
outlet = table[table["port"] == "outlet"]["pressure_kpa"].to_numpy()

plt.plot(inlet, label="inlet")
plt.plot(outlet, label="outlet")
plt.xlabel("index")
plt.ylabel("pressure_kpa")
plt.title("Pump pressure")
plt.legend()
plt.show()
```
