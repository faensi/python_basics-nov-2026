## Lesen und Filtern

`pandas` liest die ganze Tabelle. `pd.read_csv` nimmt den Pfad. Jede Spalte hat einen Namen aus der Kopfzeile.

`data/pump_pressure.csv` hat die Spalten `timestamp`, `port` und `pressure_kpa`. Die Ports heißen `inlet` und `outlet`. Mit einem Vergleich filterst du die Zeilen eines Ports.


## Zahlen als Array

`to_numpy()` wandelt eine Spalte in ein `numpy`-Array um. Auf dem Array kannst du Mittelwert und ähnliche Rechnungen mit den Methoden von `numpy` machen, zum Beispiel `mean()`.


```python
import pandas as pd
import numpy as np

table = pd.read_csv("data/pump_pressure.csv")
print(table)

inlet = table[table["port"] == "inlet"]
pressures = inlet["pressure_kpa"].to_numpy()
print("inlet array", pressures)
print("inlet mean", pressures.mean())
```
