## Aufbau der Datei

HDF5 speichert Arrays unter Namen. `data/flow_series.h5` enthält zwei Datensätze: `inlet_l_per_min` und `outlet_l_per_min`. Die Einheit ist Liter pro Minute.

`h5py.File` öffnet die Datei. Mit `"r"` liest du nur. `list(handle.keys())` zeigt die Namen. `handle["name"][()]` lädt den ganzen Datensatz als Array.


## Lesen und Mittelwert

`with` schließt die Datei wieder. Das geladene Array kannst du wie ein numpy-Array nutzen, zum Beispiel mit `mean()`.


```python
import h5py
import numpy as np

with h5py.File("data/flow_series.h5", "r") as handle:
    print("keys", list(handle.keys()))
    inlet = handle["inlet_l_per_min"][()]

print("inlet", inlet)
print("inlet mean", np.mean(inlet))
```

```python
import h5py
import numpy as np

with h5py.File("data/flow_series.h5", "r") as handle:
    inlet = handle["inlet_l_per_min"][()]
    outlet = handle["outlet_l_per_min"][()]

delta = inlet - outlet
mean_delta = np.mean(delta)
print("solution", delta)
print("solution mean_delta", mean_delta)
```
