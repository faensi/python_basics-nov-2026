## Simulierte Messwerte

Jeder Schritt hängt einen neuen Druckwert an. Der neue Wert ist der letzte Wert plus eine kleine zufällige Änderung. So entsteht eine Zufallsbewegung um den Startwert. Die Einheit ist kPa.

Das Array hält nur die letzten Werte, zum Beispiel 30. Ältere Punkte fallen weg, damit die Figur lesbar bleibt.


## Figur aktualisieren

`clear_output(wait=True)` löscht die alte Ausgabe der Zelle. Danach zeichnest du die Figur neu und wartest mit `time.sleep`. So bleibt eine dauerhafte Darstellung, die sich alle paar Sekunden erneuert.

Stoppe die Zelle mit der Unterbrechung des Kernels. Lass die Zelle im Unterricht nicht endlos laufen.

````python
import time

import matplotlib.pyplot as plt
import numpy as np
from IPython.display import clear_output

reference_kpa = 150.0
noise_kpa = 0.4
window = 30
refresh_s = 2.0
steps = 20

series = np.array([reference_kpa], dtype=float)

for _ in range(steps):
    next_value = series[-1] + np.random.normal(0.0, noise_kpa)
    series = np.append(series, next_value)
    if series.size > window:
        series = series[-window:]

    clear_output(wait=True)
    plt.plot(series)
    plt.xlabel("index")
    plt.ylabel("pressure_kpa")
    plt.title("Simulated outlet pressure")
    plt.ylim(reference_kpa - 5.0, reference_kpa + 5.0)
    plt.show()
    time.sleep(refresh_s)
````
