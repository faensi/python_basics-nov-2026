## `open` und `with`

`open(path, encoding="utf-8")` öffnet eine Datei zum Lesen. Ohne zweiten Positionsparameter ist der Modus `r` (read).

`with` sorgt dafür, dass die Datei wieder geschlossen wird, auch wenn danach ein Fehler kommt. Den geöffneten Handle nennst du oft `handle`. Nach dem `with`-Block solltest du den Handle nicht weiter benutzen.


```python
handle = open("data/notes.txt", encoding="utf-8")
print(handle)
handle.close()
```

```python
with open("data/notes.txt", encoding="utf-8") as handle:
    print(handle.closed)
print("after with", handle.closed)
```

## Ganze Datei lesen: `read`

`handle.read()` liefert den gesamten Inhalt als einen String. Geeignet für kurze Textdateien wie das Probenprotokoll.


```python
with open("data/notes.txt", encoding="utf-8") as handle:
    notes = handle.read()
print(notes[:80])
```

```python
print(len(notes))
print(notes.count("\n"))
print("Bad" in notes)
```

## Zeile für Zeile: `readline` und `for`

`readline()` liest genau eine Zeile. Am Ende steht oft das Zeichen für den Zeilenumbruch.

Eine `for`-Schleife über den Handle läuft alle restlichen Zeilen ab. Das ist der übliche Weg für längere Dateien.


```python
with open("data/notes.txt", encoding="utf-8") as handle:
    first = handle.readline()
print(repr(first))
```

```python
with open("data/notes.txt", encoding="utf-8") as handle:
    handle.readline()
    second = handle.readline()
print(repr(second))
```

```python
count = 0
with open("data/notes.txt", encoding="utf-8") as handle:
    for line in handle:
        count = count + 1
print("lines", count)
```

## Eine Zeile säubern: `strip`

`strip()` entfernt Leerzeichen und Zeilenumbrüche am Anfang und Ende. Ohne `strip` bleibt der Zeilenumbruch oft an der Zeile hängen.


```python
raw = "  bath,20.1\n"
clean = raw.strip()
print(repr(clean))
```

## CSV von Hand: `split` und `float`

`data/readings.csv` beginnt mit der Kopfzeile `timestamp,sensor,celsius`. Danach steht in jeder Zeile ein Zeitpunkt, ein Sensorname und eine Temperatur, getrennt durch Kommas.

`split(",")` zerlegt den Text in eine Liste von Teilen. Die Temperatur ist zuerst Text und wird mit `float` zur Zahl.

Leere Zeilen überspringst du mit `if not line.strip(): continue`.


```python
line = "2026-03-02T08:00:00,bath,19.94"
parts = line.split(",")
print(parts)
```

```python
timestamp, name, text_value = parts
celsius = float(text_value)
print(name, celsius)
```

```python
with open("data/readings.csv", encoding="utf-8") as handle:
    header = handle.readline().strip()
print(header)
```

## Messwerte sammeln

Du überspringst die Kopfzeile mit einem ersten `readline`. In der Schleife zerlegst du jede Zeile und hängst passende Werte an eine Liste.


```python
bath = []
with open("data/readings.csv", encoding="utf-8") as handle:
    handle.readline()
    for line in handle:
        if not line.strip():
            continue
        _timestamp, name, text_value = line.strip().split(",")
        if name == "bath":
            bath.append(float(text_value))
print("bath count", len(bath))
print("bath first", bath[0])
print("bath last", bath[-1])
```

## Schreiben: Modus `w`

`open(path, "w", encoding="utf-8")` legt die Datei neu an. Ein vorhandener Inhalt wird ersetzt.

`write` schreibt Text. Ein Zeilenumbruch entsteht nur, wenn du ihn selbst anhängst.

`data/report.txt` ist ein Ergebnis, keine Vorlage. Du darfst die Datei löschen. Der nächste Lauf schreibt sie neu.


```python
path = "data/report.txt"
text = "bath: 20.0 C\n"
print(repr(text))
```

```python
with open("data/report.txt", "w", encoding="utf-8") as handle:
    handle.write("bath: 20.0 C\n")
print("wrote one line")
```

```python
with open("data/report.txt", encoding="utf-8") as handle:
    print(handle.read())
```

## Bericht aus der Liste

Aus den gesammelten Bad-Temperaturen baust du Zeilen mit `format` und schreibst sie nacheinander.


```python
line = "bath: {:.1f} C".format(19.94)
print(line)
print(repr(line + "\n"))
```

```python
lines = []
for celsius in bath[:3]:
    lines.append("bath: {:.1f} C".format(celsius))
print(lines)
```

```python
with open("data/report.txt", "w", encoding="utf-8") as handle:
    for line in lines:
        handle.write(line + "\n")
print("wrote data/report.txt")
```

## Alles in einem Durchgang

Die nächste Zelle verbindet Lesen, Filtern und Schreiben. So sieht der Ablauf später im Skript mit F5 aus.


```python
with open("data/notes.txt", encoding="utf-8") as handle:
    notes = handle.read()
print(notes[:60])

bath = []
with open("data/readings.csv", encoding="utf-8") as handle:
    handle.readline()
    for line in handle:
        if not line.strip():
            continue
        _timestamp, name, celsius = line.strip().split(",")
        if name == "bath":
            bath.append(float(celsius))
print("bath", len(bath), bath[0], bath[-1])

lines = []
for celsius in bath:
    lines.append("bath: {:.1f} C".format(celsius))
with open("data/report.txt", "w", encoding="utf-8") as handle:
    for line in lines:
        handle.write(line + "\n")
print("wrote data/report.txt")
```
