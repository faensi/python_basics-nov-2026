## Was ist ein Modul?

Ein Modul ist eine Python-Datei mit wiederverwendbaren Namen: Funktionen, Werte, Klassen. Die Standardbibliothek kommt mit Python. Beispiele: `math`, `csv`. Sie stehen nicht in `environment.yml`.

`import` lädt das Modul einmal und stellt seine Namen bereit.


```python
import math
print(math)
print(type(math))
```

## Form 1: `import math`

Nach `import math` rufst du die Funktionen über den Modulnamen auf: `math.fabs`, `math.sqrt`. Man sieht, woher der Name kommt. Diese Form ist in diesem Kurs der Normalfall.


```python
import math
print(math.fabs(-0.4))
print(math.sqrt(9.0))
```

```python
value = -2.5
print(math.fabs(value))
print(value)
```

## Form 2: `import math as m`

`as` vergibt einen kürzeren Namen für dasselbe Modul. `m` und `math` zeigen auf dasselbe Objekt, wenn du beide Formen verwendest.


```python
import math as m
print(m.fabs(-0.4))
print(m.pi)
```

## Form 3: `from math import fabs`

Hier liegt `fabs` direkt im Skript. Du schreibst `fabs(-0.4)` ohne `math.` davor. Das ist kurz, aber der Ursprung ist weniger sichtbar.


```python
from math import fabs
print(fabs(-0.4))
print(fabs(1.2))
```

## Die drei Formen nebeneinander

Dieselbe Rechnung, drei Schreibweisen. `doctest` und `unittest` am Tag 4 gehören ebenfalls zur Standardbibliothek.


```python
import math
import math as m
from math import fabs

print("math.fabs", math.fabs(-0.4))
print("m.fabs", m.fabs(-0.4))
print("fabs", fabs(-0.4))
```

## Docstring und `help`

Der Docstring ist der Text direkt unter der Definition einer Funktion. Bei eingebauten Funktionen steht er in `__doc__`.

In der IPython-Konsole zeigt `help(math.fabs)` dieselbe Hilfe. Die Ansicht verlässt du mit `q`. Docstrings in diesem Kurs sind Englisch.


```python
import math
doc = math.fabs.__doc__
print(doc.splitlines()[0])
```

## Lokal und global

Eine Zuweisung in einer Funktion erzeugt einen lokalen Namen. Er gilt nur in der Funktion.

Einen globalen Namen darfst du in der Funktion lesen, ohne `global` zu schreiben. Erst eine Zuweisung würde einen neuen lokalen Namen erzeugen. Das hast du am Tag 2 gesehen.

Eingebaute Namen wie `len` und `open` sucht Python danach, wenn Skript und Funktion den Namen nicht selbst definiert haben.


```python
limit = 21.0
print("outside", limit)
```

```python
def show_limit():
    print("inside", limit)

show_limit()
print("still", limit)
```

```python
def first_above(values, limit):
    status = "inside"
    for value in values:
        if value > limit:
            status = "above"
            break
    return status

print(first_above([20.0, 21.4], 21.0))
print(first_above([20.0, 20.5], 21.0))
```

```python
limit = 21.0
print("limit outside", limit)
print("status", first_above([20.0, 21.4], limit))
print("limit still", limit)
```

## Messreihe mit `csv` und `math`

Du sammelst die Bad-Temperaturen über `DictReader` und berechnest die Spanne mit `math.fabs`.


```python
import csv
import math

bath = []
with open("data/readings.csv", encoding="utf-8", newline="") as handle:
    reader = csv.DictReader(handle)
    for row in reader:
        if row["sensor"] == "bath":
            bath.append(float(row["celsius"]))
print("bath count", len(bath))
print("bath spread", round(math.fabs(max(bath) - min(bath)), 1))
```
