## Warum eine zweite Datei

`measurement.py` enthält den Mittelwert, die Abweichung, die Toleranzprüfung und die Umrechnung von Grad Celsius nach Kelvin. Das sind dieselben Rechnungen wie an den vorigen Tagen, jetzt an einem Ort, den mehrere Skripte nutzen können.

Der Dateiname ohne `.py` ist der Name im `import`. Die Datei `measurement.py` wird also als `measurement` importiert.


```python
from pathlib import Path
print(Path("tag3/measurement.py").exists())
print(Path("tag3/measurement.py").name)
```

## Modul finden

Liegt das Arbeitsverzeichnis auf dem Kursordner, findet Python `measurement.py` nicht von allein. Die nächste Zelle legt den Ordner `tag3` in den Suchpfad. Startest du das Skript mit F5 aus Spyder, reicht oft schon der Ordner der Skriptdatei.


```python
import sys
from pathlib import Path

module_dir = Path("tag3")
if not (module_dir / "measurement.py").exists():
    module_dir = Path(".")
sys.path.insert(0, str(module_dir.resolve()))
print(module_dir.resolve())
```

## Import des eigenen Moduls

Nach dem Import ist `measurement` ein Modulobjekt. Die Funktionen erreichst du mit Punkt: `measurement.mean`.


```python
import measurement
print(measurement)
print(type(measurement))
```

```python
print(measurement.mean)
print(callable(measurement.mean))
```

## `mean`

`measurement.mean` erwartet eine nicht leere Zahlenliste und gibt den Mittelwert zurück.


```python
readings = [20.0, 21.0, 22.0]
print(measurement.mean(readings))
```

```python
print(measurement.mean([19.0, 21.0]))
print(measurement.mean([20.0]))
```

## `deviation` und `within_tolerance`

`deviation` ist Anzeige minus Referenz. `within_tolerance` prüft, ob der Betrag der Abweichung höchstens so groß wie die Toleranz ist.


```python
print(measurement.deviation(21.4, 21.0))
print(round(measurement.deviation(21.4, 21.0), 1))
```

```python
print(measurement.within_tolerance(21.4, 21.0, 0.5))
print(measurement.within_tolerance(22.0, 21.0, 0.5))
```

## `celsius_to_kelvin`

Die Umrechnung addiert `273.15`. Das Ergebnis ist eine Zahl in Kelvin.


```python
print(measurement.celsius_to_kelvin(20.0))
print(measurement.celsius_to_kelvin(0.0))
```

## Docstring und Hilfe

Der englische Text unter `def mean` in `measurement.py` ist der Docstring. `measurement.mean.__doc__` gibt ihn aus. In der IPython-Konsole zeigt `help(measurement.mean)` dieselbe Hilfe. Die Ansicht verlässt du mit `q`.


```python
print(measurement.mean.__doc__)
```

```python
print(measurement.deviation.__doc__)
print(measurement.within_tolerance.__doc__)
```

## Kurzer Alias

Wie bei der Standardbibliothek kannst du dem Modul einen kürzeren Namen geben. Die Funktionen bleiben dieselben.


```python
import measurement as m
print(m.mean([20.0, 22.0]))
print(m.celsius_to_kelvin(20.0))
```

## Alles in einem Durchgang

Die nächste Zelle ruft die vier Funktionen nacheinander auf und zeigt den Docstring von `mean`.


```python
import measurement

readings = [20.0, 21.0, 22.0]
print("mean", measurement.mean(readings))
print("deviation", round(measurement.deviation(21.4, 21.0), 1))
print("within", measurement.within_tolerance(21.4, 21.0, 0.5))
print("kelvin", measurement.celsius_to_kelvin(20.0))
print("doc:", measurement.mean.__doc__)
```
