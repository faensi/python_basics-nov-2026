## Was ist Debugging?

Debugging heißt: herausfinden, warum ein Programm nicht das tut, was du erwartest.

Zwei typische Fälle:

1. Das Programm bricht ab. Python zeigt eine Fehlermeldung (Traceback).
2. Das Programm läuft durch, aber das Ergebnis ist falsch. Dann brauchst du Zwischenwerte.


## Eine Fehlermeldung lesen

Führe die nächste Zelle aus. Unten steht der Traceback. Lies ihn von unten nach oben:

- die letzte Zeile nennt den Fehlertyp und die kurze Meldung
- darüber steht die Zeile in deinem Code, die den Fehler ausgelöst hat


```python
numbers = [10.0, 20.0, 30.0]
print(numbers[5])
```

## Typische Fehlermeldungen

Die nächsten Zellen erzeugen absichtlich Fehler. Lies jeweils den Traceback. Danach kannst du die Zelle ändern oder überspringen.


### `NameError`

Der Name wurde noch nicht definiert, oder er ist falsch geschrieben.


```python
print(temperature)
```

### `TypeError`

Ein Vorgang passt nicht zum Typ. Hier addierst du Text und Zahl.


```python
text = "20.1"
print(text + 1)
```

### `ValueError`

Der Typ passt, der Inhalt nicht. `"room"` ist kein Zahlenwert.


```python
print(float("room"))
```

### `IndexError`

Der Index liegt außerhalb der Liste. Die Liste hat die Indizes 0, 1 und 2.


```python
readings = [19.8, 20.1, 19.9]
print(readings[3])
```

### `ZeroDivisionError`

Du teilst durch null.


```python
count = 0
print(10 / count)
```

### `KeyError`

Der Schlüssel fehlt im Dictionary.


```python
row = {"sensor": "bath", "celsius": 20.1}
print(row["timestamp"])
```

### `FileNotFoundError`

Die Datei liegt nicht am angegebenen Pfad. Oft ist das Arbeitsverzeichnis falsch.


```python
with open("missing_file.txt", encoding="utf-8") as handle:
    print(handle.read())
```

## Mit `print` Zwischenwerte prüfen

Viele Fehler findest du ohne Debugger. Du gibst die wichtigen Namen aus und vergleichst sie mit dem, was du erwartest.


```python
total = 0.0
for value in [1.0, 2.0, 3.0]:
    total = total + value
    print("value", value, "total", total)
print("end", total)
```

## Off-by-one: einer zu weit

Die Liste hat drei Werte. Die Indizes sind 0, 1 und 2. Die Länge ist 3.

`readings[len(readings)]` greift einen Schritt zu weit. Das ist ein Off-by-one-Fehler: die Rechnung ist fast richtig, aber um eins verschoben.


```python
readings = [19.8, 20.1, 19.9]
print("length", len(readings))
print("last valid index", len(readings) - 1)
print("last value", readings[len(readings) - 1])
```

```python
readings = [19.8, 20.1, 19.9]
print(readings[len(readings)])
```

## Denselben Fehler mit `print` finden

Vor dem Zugriff gibst du den Index aus. So siehst du, dass `index` gleich 3 ist, obwohl der letzte gültige Index 2 ist.


```python
readings = [19.8, 20.1, 19.9]
index = len(readings)
print("index", index)
print("allowed", 0, "to", len(readings) - 1)
print(readings[index])
```

## Die Korrektur

Der letzte Wert liegt bei `len(readings) - 1`, oder kürzer bei `readings[-1]`.


```python
readings = [19.8, 20.1, 19.9]
print(readings[len(readings) - 1])
print(readings[-1])
```

## Fehler abfangen: `try` und `except`

Manchmal weißt du, dass ein Schritt scheitern kann, und das Programm soll trotzdem weiterlaufen. Dann fängst du den Fehler ab.

In Python heißt das `try` / `except` (nicht `catch`).

Sinnvoll ist das vor allem bei äußeren Ursachen: Datei fehlt, Pfad falsch, Eingabe ungültig. Nicht sinnvoll ist es, jeden Programmierfehler zu verstecken. Einen Off-by-one in deinem Index solltest du korrigieren, nicht abfangen.


### Datei lesen mit Absicherung

Ohne `try` bricht `FileNotFoundError` die Zelle ab. Mit `except FileNotFoundError` gibst du eine klare Meldung aus und gehst weiter.


```python
path = "data/notes.txt"
try:
    with open(path, encoding="utf-8") as handle:
        text = handle.read()
    print(text[:40])
except FileNotFoundError:
    print("file not found:", path)
```

```python
path = "data/does_not_exist.txt"
try:
    with open(path, encoding="utf-8") as handle:
        text = handle.read()
    print(text)
except FileNotFoundError:
    print("file not found:", path)
```

### Nur den erwarteten Fehler fangen

Fange den Fehlertyp, den du meinst. `except FileNotFoundError` ist enger und klarer als ein nacktes `except:`.


```python
value = "20.1"
try:
    number = float(value)
    print(number)
except ValueError:
    print("not a number:", value)
```

```python
value = "room"
try:
    number = float(value)
    print(number)
except ValueError:
    print("not a number:", value)
```

### Wann `try` / `except` Sinn hat

- Datei öffnen: Pfad kann fehlen
- Text in Zahl wandeln: Eingabe kann ungültig sein
- Netzwerk oder Geräte (später): Verbindung kann scheitern

Weniger sinnvoll:

- Indexfehler in einer festen Liste, die du selbst gebaut hast
- Tippfehler in Variablennamen
- Logikfehler wie Off-by-one
