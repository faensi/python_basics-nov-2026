## `__init__` und `self`

`__init__` richtet eine neue Instanz ein. Python ruft sie beim Erzeugen automatisch auf.

`self` ist die neue Instanz. `self.sensor` und `self.celsius` sind Attribute **dieser** Instanz.


```python
class Reading:
    """One temperature reading from a named sensor."""

    def __init__(self, sensor, celsius):
        self.sensor = sensor
        self.celsius = celsius

bath = Reading("bath", 20.1)
print(bath.sensor)
print(bath.celsius)
```

## Zwei Instanzen bleiben getrennt

`bath` und `room` sind zwei Objekte. Änderst du `room.celsius`, bleibt `bath.celsius` unverändert.


```python
bath = Reading("bath", 20.1)
room = Reading("room", 21.5)
print("bath", bath.sensor, bath.celsius)
print("room", room.sensor, room.celsius)
room.celsius = 21.6
print("bath still", bath.celsius)
print("room now", room.celsius)
```

## Methoden

Eine Methode ist eine Funktion in der Klasse. Der erste Parameter heißt `self` und ist die Instanz.

Beim Aufruf `bath.kelvin()` übergibt Python `bath` als `self`. Du schreibst `self` nicht selbst in die Klammern.


```python
class Reading:
    """One temperature reading from a named sensor."""

    def __init__(self, sensor, celsius):
        self.sensor = sensor
        self.celsius = celsius

    def kelvin(self):
        """Return this reading converted to kelvin."""
        return self.celsius + 273.15

bath = Reading("bath", 20.1)
print(bath.kelvin())
```

```python
class Reading:
    """One temperature reading from a named sensor."""

    def __init__(self, sensor, celsius):
        self.sensor = sensor
        self.celsius = celsius

    def deviation(self, reference):
        """Return this reading minus the reference value."""
        return self.celsius - reference

    def kelvin(self):
        """Return this reading converted to kelvin."""
        return self.celsius + 273.15

bath = Reading("bath", 20.1)
print(round(bath.deviation(20.0), 1))
print(bath.kelvin())
```

## Alles in einem Durchgang

Die nächste Zelle zeigt Klasse, zwei Instanzen, Methodenaufrufe und getrennte Attribute.


```python
class Reading:
    """One temperature reading from a named sensor."""

    def __init__(self, sensor, celsius):
        self.sensor = sensor
        self.celsius = celsius

    def deviation(self, reference):
        """Return this reading minus the reference value."""
        return self.celsius - reference

    def kelvin(self):
        """Return this reading converted to kelvin."""
        return self.celsius + 273.15

bath = Reading("bath", 20.1)
room = Reading("room", 21.5)
print("class", Reading.__name__)
print("instance class", type(bath).__name__)
print("bath sensor", bath.sensor)
print("room sensor", room.sensor)
print("bath deviation", round(bath.deviation(20.0), 1))
print("bath kelvin", bath.kelvin())
room.celsius = 21.6
print("bath celsius still", bath.celsius)
print("room celsius", room.celsius)
```
