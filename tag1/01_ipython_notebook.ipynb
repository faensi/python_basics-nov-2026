## IPython

IPython ist eine interaktive Shell, und du schreibst einen Ausdruck und siehst das Ergebnis sofort. Die Konsole in Spyder ist eine IPython-Konsole.

- **F5** führt ein geöffnetes Skript aus.
- **F9** führt die markierte Zeile in der Konsole aus.

Ein Notebook-Kernel ist derselbe Gedanke: ein laufender Python-Prozess, der sich Namen merkt. Starte den Kernel neu, wenn alte Namen stören. In Spyder startest du dafür die IPython-Konsole neu.


## Zellen

Ein Notebook speichert Text und Code in Zellen.

- Eine Markdown-Zelle enthält die Erklärung, so wie dieser Abschnitt.
- Eine Code-Zelle enthält Python.

Es zählt die Reihenfolge, in der du die Zellen ausführst, nicht nur die Reihenfolge auf der Seite. Eine Zelle kann einen Namen aus einer früheren Zelle benutzen. Fehlt der Name, war die frühere Zelle noch nicht ausgeführt.


## Erster Code

Die letzte Zeile einer Zelle zeigt ihren Wert auch ohne `print`. `reading` steht deshalb nach den Druckzeilen noch einmal allein.


````python
print("Hello from a notebook cell")
print(19.5 + 1.5)
reading = 20.1
reading
````

## Eine kurze Messreihe

Die Liste hält fünf Temperaturen in Grad Celsius. Der Index beginnt bei 0, wie im Skript der nächsten Einheiten.


````python
readings = [19.8, 20.1, 19.9, 20.0, 20.2]
print("count", len(readings))
print("first", readings[0])
````

## Übung

Lege in der nächsten Zelle drei eigene Variablen an. Übernimm nicht die Zahlen 19,5 und 1,5 aus der Zelle weiter oben und nicht die Temperatur 20,1.

- `place` ist ein Text für einen Ort.
- `morning` ist eine Temperatur in Grad Celsius mit Nachkommastelle.
- `evening` ist eine zweite Temperatur mit Nachkommastelle.

Ersetze den leeren Text und die Nullen. Gib den Ort aus und danach die Summe der beiden Temperaturen.


````python
place = ""
morning = 0.0
evening = 0.0
print(place)
print(morning + evening)
````

## Lösung

```python
place = "hall"
morning = 18.5
evening = 19.0
print(place)
print(morning + evening)
```

Andere Orte und andere Temperaturen mit Nachkommastelle sind ebenfalls richtig.
