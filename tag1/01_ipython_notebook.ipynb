## IPython

IPython ist eine interaktive Shell. Sie schreiben einen Ausdruck und sehen das Ergebnis sofort. Die Konsole in Spyder ist eine IPython-Konsole.

- **F5** führt ein geöffnetes Skript aus.
- **F9** führt die markierte Zeile in der Konsole aus.

Ein Notebook-Kernel ist derselbe Gedanke: ein laufender Python-Prozess, der sich Namen merkt. Starten Sie den Kernel neu, wenn alte Namen stören. In Spyder starten Sie dafür die IPython-Konsole neu.


## Zellen

Ein Notebook speichert Text und Code in Zellen.

- Eine Markdown-Zelle enthält die Erklärung, so wie dieser Abschnitt.
- Eine Code-Zelle enthält Python.

Es zählt die Reihenfolge, in der Sie die Zellen ausführen, nicht nur die Reihenfolge auf der Seite. Eine Zelle kann einen Namen aus einer früheren Zelle benutzen. Fehlt der Name, war die frühere Zelle noch nicht ausgeführt.
